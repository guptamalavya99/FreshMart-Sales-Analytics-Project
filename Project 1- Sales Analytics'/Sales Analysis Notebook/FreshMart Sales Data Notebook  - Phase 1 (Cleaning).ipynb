```python
import re

def smart_parse_dates(s):

    s = s.astype(str)
    out = pd.Series(pd.NaT, index=s.index)
    patterns = [
        (r'^\d{4}-\d{2}-\d{2}$',        '%Y-%m-%d'),
        (r'^\d{2}/\d{2}/\d{4}$',        '%m/%d/%Y'),
        (r'^\d{2}-\d{2}-\d{4}$',        '%d-%m-%Y'),
        (r'^\d{2} [A-Za-z]{3} \d{4}$',  '%d %b %Y'),
        (r'^\d{4}/\d{2}/\d{2}$',        '%Y/%m/%d'),
    ]
    for pat, fmt in patterns:
        mask = s.str.match(pat, na=False)
        out.loc[mask] = pd.to_datetime(s[mask], format=fmt, errors='coerce')
    return out
```
